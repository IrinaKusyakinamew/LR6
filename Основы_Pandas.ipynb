```python
df.head(10)
```

Выводим часть таблицы как объект series:

```python
df['city']
```

Выводим часть датафрейма, заключив название столбца в дополнительные квадратные скобки:

```python
df[['city']]
```

Для обращения к нескольким столбцам таблицы необходимо указывать их в ДВОЙНЫХ КВАДРАТНЫХ СКОБКАХ. Обращение в одинарных скобках допустимо только к ОДНОМУ столбцу. Остальные варианты приведут к ошибкам

```python
df[['city', 'gender', 'experience']]
```
