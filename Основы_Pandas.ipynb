# Фильтрация по квантилю

Иногда нужно убрать (сделать "выброс" значения) либо очень большие, либо очень маленькие значения переменных. Это делается для улучшения качества работы алгоритмов машинного обучения

Получить значение для процентиля можно при помощи квантиля (оборачиваем во float, чтобы выводилось не np.float64(0.74), а просто 0,74

```python
q = float(df['city_development_index'].quantile(0.25))
```

```python
q
```

Теперь можно вывести 25% датасета, значения по ключевым показателям столбцов в которых меньше 0,74

```python
df[df['city_development_index'] < q]
```

Сделаем проверку, что выведенная таблица - это действительно 25% датасета

```python
len(df[df['city_development_index'] < q]) / len(df)
```

Теперь можно фильтровать данные, получая нужные значения квантилей

```python
q = float(df['city_development_index'].quantile(0.05))
```

Вывод оригинальной гистограммы

```python
df['city_development_index'].hist()
```

Вывод гистограммы с отфильтрованными данными. Теперь значения будут начинаться с 0.6, а не 0.4 (по оси абсцисс)

```python
df[df['city_development_index'] >= q]['city_development_index'].hist()
```

Еще один пример использования фильтрации квантилей

```python
q1 = float(df['city_development_index'].quantile(0.05))
q2 = float(df['city_development_index'].quantile(0.95))
```

```python
df[df['city_development_index'].between(q1, q2)]['city_development_index'].hist()
```

# Сводники для непрерывных переменных

Если попытаться построить сводную таблицу для непрервных переменных так же, как и для категориальных, то получим довольно большой объем данных, проанализировать который будет довольно трудно

```python
df.groupby('city_development_index', dropna=False)['target'].agg(['count', 'mean'])
```

Поэтому прежде чем строить сводную таблицу по непрерывной переменной, нужно предварительно разбить ее на интервалы

Посредством функции cut указываем переменную, которую хотим разбить на интервалы, а также количество интервалов для разбиения. В итоге указанная переменная разбивается на 5 равных по длине интервалов

```python
pd.cut(df['city_development_index'], 5)
```

Сохраним полученное значение в новой переменной city_development_index_group

```python
df['city_development_index_group'] = pd.cut(df['city_development_index'], 5)
```

Затем вместо непрерывной переменной city_development_index для составления сводной таблицы используем разбитую на равные интервалы переменную city_development_index_group. В итоге получаем наглядную сводную таблицу

```python
df.groupby('city_development_index_group', dropna=False)['target'].agg(['count', 'mean'])
```

Полученные значения в сводной таблице можно сравнить с гистограммой по city_development_index

```python
df['city_development_index'].hist()
```

Чтобы сделать группы, равномерные не просто по интервалу, а еще и по какому-то определенному признаку (к примеру, по количеству людей), используют qcut. Этот метод использует квантили для разбиения данных

Разобьем непрерывную переменную city_development_index на 5 интервалов по количеству людей в этих интервалах

```python
df['city_development_index_group_q'] = pd.qcut(df['city_development_index'], 5)
```

Ошибка возникает из-за большого количества повторяющихся значений. Разбить данные на 5 одинаковых групп невозможно, если большинство записей сосредоточено в определенном интервале

Это происходит, когда распределение значений очень "скученное" — например, много одинаковых значений в одном месте. Если в расчёте квантилей получается, что 4-й и 5-й квантиль совпадают (одинаковое число), то Pandas не может построить нужное число интервалов.

Чтобы решить проблему, используют duplicates='drop' для удаления повторяющихся данных

duplicates управляет тем, что делать, если расчёт квантилей дал одинаковые границы интервалов:

duplicates='raise' (по умолчанию) → выбросит ошибку, если границы неуникальны.

duplicates='drop' → уберёт дубликаты границ и создаст меньше групп, чем запрошено.

```python
df['city_development_index_group_q'] = pd.qcut(df['city_development_index'], 5, duplicates='drop')
```

Выведем получившуюся сводную таблицу

В результате данные разобьются не на 5 интервалов, а на 4.

```python
df.groupby('city_development_index_group_q', dropna=False)['target'].agg(['count', 'mean'])
```

Проверим контрольную сумму количества записей

```python
t = df.groupby('city_development_index_group_q', dropna=False)['target'].agg(['count', 'mean'])
```

```python
int(t['count'].sum())
```

Построим график по среднему значению mean (процент желающих сменить работу)

```python
t['mean'].plot()
```

Таким образом, чем больше развит город, тем ниже процент людей, желающих найти новую работу

Метод cut также позволяет создать кастомные интервалы - те, которые можно задать вручную, причем они не обязательно должны быть одинаковыми

```python
df['city_development_index_group_custom'] = pd.cut(df['city_development_index'], [0, 0.1, 0.2, 0.5, float('inf')])
```

Построим сводную таблицу

```python
df.groupby('city_development_index_group_custom', dropna=False)['target'].agg(['count', 'mean'])
```

Если нужно включить в границу интервала 0, используем доп. интервал с -беск

```python
df['city_development_index_group_custom'] = pd.cut(df['city_development_index'], [-float('inf'), 0, 0.1, 0.2, 0.5, float('inf')])
```

```python
df.groupby('city_development_index_group_custom', dropna=False)['target'].agg(['count', 'mean'])
```

# Группировка в категориальных переменных

```python
df.groupby('education_level', dropna=False)['target'].agg(['count', 'mean'])
```

Объединение разных категорий осуществляется несколькими способами:
1.  Написать функцию, которая будет обрабатывать значение в столбце, и применить ее через метод apply
2. Через replace
3. Через словарь

```python
def education_group(x):
  if x in ['High School', 'Primary School']:
    return 'School'
  if x in ['Masters', 'Phd']:
    return 'Masters_and_phd'
  return x
```

Применим функцию к нужному столбцу при помощи apply

```python
df['education_level'].apply(education_group)
```

Сохраним полученную группировку в новый столбец датафрейма

```python
df['new_education_level'] = df['education_level'].apply(education_group)
```

Построим сводную таблицу для нового столбца с новой группировкой

```python
df.groupby('new_education_level', dropna=False)['target'].agg(['count', 'mean'])
```

Сделать группировку можно без использования функций, для этого к нужному столбцу следует применить метод replace (один или несколько раз)

```python
df['education_level'].replace('Primary School', 'School').replace('High School', 'School')
```

Также можно использовать словарь, в котором будут прописаны все нужные группировки. Нужно прописать замены для всех групп, иначе для не указанных групп будут выведены NaN

```python
d = {'Primary School': 'School', 'High School': 'School'}
```

А потом применить его к нужному столбцу при помощи map

# Функция с несколькими столбцами

В функции также можно использовать нескольких столбцов: к примеру, когда нужно заполнить один столбец в зависимости от условия по другому столбцу. Такая функция будет применяться уже ко всему датафрейму

К примеру, увеличим количество часов в столбце training_hours на 1000 для группы людей phd. В данном случае "для группы людей phd" - условие, которое будем проверять по столбцу education_level

```python
def new_training_hours(row):
  if row['education_level'] == 'Phd':
    return row['training_hours'] + 1000
  return row['training_hours']
```

```python
df.apply(new_training_hours, axis=1)
```

Создадим новый столбец датафрейма и сохраним в него полученную группировку

```python
df['new_training_hours'] = df.apply(new_training_hours, axis=1)
```
